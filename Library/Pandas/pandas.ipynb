### Creating lables

```python
import pandas as pd
a = [1, 7, 2]
myvar = pd.Series(a, index = ["x", "y", "z"])
print(myvar)
```

### Key/value object as series 

```python
import pandas as pd 
calories = {"day1": 890, "day2": 432, "day3": 999}
myvar = pd.Series(calories)
print(myvar)
```

```python
import pandas as pd 
calories = {"day1": 232, "day2": 234, "day3": 987}
myvar = pd.Series(calories, index= ["day1", "day2"])
print(myvar["day2"])
```

```python
import pandas as pd
data = {
  "Name" : ["John", "Anna", "Peter", "Linda"],
  "Location" : ["New York", "Paris", "Berlin", "London"],
  "Age" : [24, 13, 53, 33]
}

df = pd.DataFrame(data)
filtered = df[df["Age"]>20]
print(filtered)
```

```python
filtered = df[df["Location"] != "India"]
print(filtered)
```

multiple conditions

```python
filtered = df[(df["Name"] == "Anna") & (df["Age"] > 30)]
print(filtered)
```

adding new column

```python
df["Marks"] = [85, 92, 78, 90]
print(df)
```

```python
df.loc[0, "Marks"] = 95
print(df)
```

```python
print(df.columns)
```

```python
df["Location"] = ["NY", "Paris", "Berlin", "London"]
```

```python
print(df.columns)
```

```python
df.drop("Marks", axis=1, inplace=True)
print(df)
```

### Handling missing values
#### Check missing values

```python
print(df.isna())
```

#### Remove Missing Values

```python
df.dropna()
```

```python
df["Gender"] = ["M", "F", "M", "F"]
print(df)
```

```python
df[df["Age"]>20]
```

```python
df.query("Age > 20 and Gender == 'M'")
```

```python
df["Age_plus_10"] = df["Age"].apply(lambda x: x + 10)
print(df)
```

```python
df.fillna(0)
```

### GroupBy

```python
data = {
    "Department": ["IT", "IT", "HR", "HR"],
    "Salary": [50000, 60000, 45000, 47000]
}

df = pd.DataFrame(data)

print(df.groupby("Department").median())
```
