```python
#Store changes
df = df.dropna()
```

```python
#fill with mean
df["Age"]= df["Age"].fillna(df["Age"].mean())
```

```python
#fill with median 
df["Age"]= df["Age"].fillna(df["Age"].median())
```

```python
df["Gender"] = df["Gender"].fillna(df["Gender"].mode()[0])
```

```python
df["Gender"].mode()
```

```python
df["Gender"].mode()[0]
```

```python
categorical_columns = [
  "Gender", "Blood Type", "Medical Condition", 
  "Doctor", "Hospital", "Insurance Provider", 
  "Admission Type", "Medication", "Test Results"
]

for col in categorical_columns:
  df[col] = df[col].fillna(df[col].mode()[0])
```

```python
#forward fill
df.fillna(method= "ffill")
```

```python
#backward fill
df.fillna(method= "bfill")
```

```python
patients = {
    "Patient_ID": [101, 102, 103, 104, 105],
    "Name": ["John Smith", "Alice Brown", "David Lee", "Emma Wilson", "Michael Scott"],
    "Age": [25, 32, 45, 28, 50],
    "Gender": ["Male", "Female", "Male", "Female", "Male"],
    "Hospital": ["Apollo", "Fortis", "Apollo", "Max", "Fortis"]
}
```

```python
df_patients = pd.DataFrame(patients)
```

```python
billing = {
    "Patient_ID": [101, 102, 103, 104, 105],
    "Doctor": ["Dr. Sharma", "Dr. Patel", "Dr. Gupta", "Dr. Khan", "Dr. Singh"],
    "Billing Amount": [12000, 8500, 15000, 7000, 20000],
    "Insurance": ["Yes", "No", "Yes", "Yes", "No"]
}
```

```python
df_billing = pd.DataFrame(billing)
```

```python
merged = pd.merge(df_patients, df_billing, on="Patient_ID")
```

```python
print(merged)
```

```python
#concat joins datframe
#horizontal (adds columns)
pd.concat([df_patients, df_billing], axis= 1)
```

```python
#vertical (adds rows)
pd.concat([df_patients, df_billing])
```
