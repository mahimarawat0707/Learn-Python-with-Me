### Pandas series
Series is like a column

```python
import pandas as pd 
a = [1,4,5]
myvar = pd.Series(a)
print(myvar)
```

### Labels

```python
print(myvar[2])
```

### Creating lables

```python
import pandas as pd
a = [1, 7, 2]
myvar = pd.Series(a, index = ["x", "y", "z"])
print(myvar)
```

```python
print(myvar["z"])
```

### Key/value object as series 

```python
import pandas as pd 
calories = {"day1": 890, "day2": 432, "day3": 999}
myvar = pd.Series(calories)
print(myvar)
```

```python
import pandas as pd 
calories = {"day1": 232, "day2": 234, "day3": 987}
myvar = pd.Series(calories, index= ["day1", "day2"])
print(myvar["day2"])
```

### Dataframes
 Data sets in Pandas are usually multi-dimensional tables

```python
import pandas as pd 
data = {
  "calories": [420, 380, 390],
  "duration": [50, 40, 45]
}

myvar = pd.DataFrame(data)

print(myvar)
```
