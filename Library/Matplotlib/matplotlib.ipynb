```python
import matplotlib.pyplot as plt

subjects = ["Python", "SQL", "Excel", "Power BI"]
students = [40, 35, 45, 30]

plt.bar(subjects, students)

plt.title("Students in Different Subjects")
plt.xlabel("Subjects")
plt.ylabel("Number of Students")

plt.show()
```

```python
import matplotlib.pyplot as plt

subjects = ["Python", "SQL", "Excel", "Power BI"]
students = [40, 35, 45, 30]

plt.pie(students, labels=subjects, autopct="%1.1f%%")

plt.title("Subject-wise Distribution")

plt.show()
```
